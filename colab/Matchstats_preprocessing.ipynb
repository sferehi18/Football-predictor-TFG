## Union de datos

Para unir nuestros datos provvenientes de distintas fuentes realizamos un renombrado a las columnas correspondientes con `equipo local, equipo visitante y fecha` y una vez hecho esto podremos hacer merge de ambos dataframes por dichas columnas.

```python
# 1. Asegurar formato Datetime y ORDENAR (Crítico para merge_asof)
df_football_co['date'] = pd.to_datetime(df_football_co['date'])
df_understat['date'] = pd.to_datetime(df_understat['date'])

df_football_co = df_football_co.sort_values('date')
df_understat = df_understat.sort_values('date')

# 2. Normalizar nombres de columnas para que coincidan en la union
df_football_co = df_football_co.rename(columns={'HomeTeam': 'home_team', 'AwayTeam': 'away_team'})

df_merged = pd.merge_asof(
    df_understat,
    df_football_co,
    on='date',
    by=['home_team', 'away_team'],
    tolerance=pd.Timedelta(days=3),
    direction='backward'  # ✅ Solo mira hacia atrás en el tiempo
)
```

## Sección 2: Ingeniería de Características (Estado de Forma)
Para que el modelo pueda predecir un partido futuro, necesita conocer el "estado de forma" con el que llegan los equipos. En lugar de usar estadísticas acumuladas de toda la temporada, calculamos promedios móviles de los últimos 5 partidos (`avg5`) y tendencias a corto plazo (`trend`).

 **Control de Fuga de Datos (Data Leakage):**
Es de vital importancia el uso de la función `.shift(1)` en Pandas antes de calcular la media móvil. Esto garantiza que las estadísticas que el modelo evalúa para el partido de "hoy" se basen estrictamente en partidos anteriores, impidiendo que la red neuronal "vea el futuro" y memorice el resultado actual.

## Calculo de diferenciales

Según varias fuentes y la tendencia actual en cuanto a predicción de resultados en deportes del tipo 1x2 se refiere. Los modelos funcionan mejor al calcular la diferencia entre las estadisticas de los equipos.

```python

import matplotlib.pyplot as plt
from sklearn.feature_selection import f_classif
import numpy as np
import pandas as pd

# Extraemos las columnas numericas de nuestro dataframe de estadisticas unido por una parte
X = df_merged[numeric_columns].copy()

#Extraemos las columnas que aportan el contexto del partido
df_context = df_merged[['fecha', 'hora']].copy()
y = df_merged['resultado_final'].copy()

calculated_features = pd.DataFrame(index=X.index)



# Calculamos diferenciales entre estadisticas de local y visitante de cada partido

calculated_features['diff_xg'] = X['xg_loc'] - X['xg_vis']
calculated_features['diff_tiros_puerta'] = (
    X['tiros_puerta_loc'] - X['tiros_puerta_vis']
)

calculated_features['diff_tiros'] = (
    X['tiros_loc'] - X['tiros_vis']
)

calculated_features['diff_pases_area'] = (
    X['pases_area_loc'] - X['pases_area_vis']
)

calculated_features['diff_xpts'] = (
    X['xpts_loc'] - X['xpts_vis']
)

calculated_features['diff_corners'] = (
    X['corners_loc'] - X['corners_vis']
)

calculated_features['diff_faltas'] = (
    X['faltas_loc'] - X['faltas_vis']
)

calculated_features['diff_amarillas'] = (
    X['amarillas_loc'] - X['amarillas_vis']
)



# 5. PRESIÓN / ESTILO

# Más positivo = local presiona más
calculated_features['diff_ratio_presion'] = (
    X['ppda_vis'].replace(0, np.nan) -
    X['ppda_loc'].replace(0, np.nan)
)



# 6. FEATURES COMPUESTAS

# Dominio ofensivo general
calculated_features['diff_dominio_ofensivo'] = (
    (X['xg_loc'] + X['tiros_loc'] + X['pases_area_loc'])
    -
    (X['xg_vis'] + X['tiros_vis'] + X['pases_area_vis'])
)

# Agresividad ofensiva simple
calculated_features['diff_agresividad_ofensiva'] = (
    (X['corners_loc'] + X['tiros_loc'])
    -
    (X['corners_vis'] + X['tiros_vis'])
)

# Disciplina (roja pesa más)
calculated_features['diff_disciplina'] = (
    (X['amarillas_vis'] + 2 * X['rojas_vis'])
    -
    (X['amarillas_loc'] + 2 * X['rojas_loc'])
)


# Limpieza de nulos (si los hay)

calculated_features = calculated_features.replace(
    [np.inf, -np.inf], np.nan
)

# Mejor usar mediana que cero
calculated_features = calculated_features.fillna(
    calculated_features.median(numeric_only=True)
)
```

## Analisis Anova

```python

f_values, p_values = f_classif(calculated_features, y)

anova_df = pd.DataFrame({
    'feature': calculated_features.columns,
    'f_value': f_values,
    'p_value': p_values
}).sort_values(by='f_value', ascending=False)


# VISUALIZACIÓN

plt.figure(figsize=(15, 7))
plt.bar(anova_df['feature'], anova_df['f_value'])
plt.xticks(rotation=90)
plt.axhline(
    y=3.0,
    color='r',
    linestyle='--',
    label='Umbral orientativo'
)
plt.xlabel('Features')
plt.ylabel('F-value')
plt.title('Importancia ANOVA para Predicción 1X2')
plt.legend()
plt.tight_layout()
plt.show()


# RESULTADOS

print("Top 15 variables con mayor poder predictivo:")
print(anova_df.head(15))
```

# Calculo de medias por Equipo

Para que el modelo pueda predecir un partido futuro, necesita conocer el "estado de forma" con el que llegan los equipos. En lugar de usar estadísticas acumuladas de toda la temporada, calculamos promedios móviles de los últimos 5 partidos (`avg5`). Para realizar estos calculos ...

```python
import pandas as pd

# Unimos todo lo calculado con nuestro dataframe mergeado al inicio
df_complete = pd.concat([df_merged.loc[:,['equipo_loc','equipo_vis']], X, calculated_features, df_context,target], axis=1)

# Extraemos las columnas de nuestro dataframe
columnas = df_complete.columns.tolist()
comun_cols = ['fecha', 'hora', 'resultado_final']

# SEPARACIÓN LOCAL
# Filtramos columnas que sean 'comunes', que tengan '_loc' o que sean 'diff_' (calculados anteriormente)
df_local = df_complete[[c for c in columnas if '_loc' in c or c in comun_cols or c.startswith('diff_')]].copy()


# Inversión de signos (Para el local es * 1, no cambia, pero lo dejamos por estructura)
# diff_cols ya se llaman 'diff_xg', 'diff_tiros', etc.
diff_cols = [c for c in df_local.columns if c.startswith('diff_')]

# Quitamos el sufijo '_loc'
df_local.columns = [c.replace('_loc', '') for c in df_local.columns]
df_local = df_local.rename(columns={'equipo': 'Equipo', 'fecha': 'Fecha'})


# Establecemos una columna que nos indica que en los partidos de este dataframe los equipos jugaban como locales
df_local['es_local'] = 1



# SEPARACIÓN VISITANTE
df_visita = df_complete[[c for c in columnas if '_vis' in c or c in comun_cols or c.startswith('diff_')]].copy()

# INVERSIÓN CRÍTICA: Aquí las 'diff_' se vuelven negativas para el visitante
diff_cols_vis = [c for c in df_visita.columns if c.startswith('diff_')]
df_visita.loc[:, diff_cols_vis] = df_visita.loc[:, diff_cols_vis] * -1

# RENOMBRAR: Quitamos '_vis' para que los nombres coincidan con los del local
df_visita.columns = [c.replace('_vis', '') for c in df_visita.columns]
df_visita = df_visita.rename(columns={'equipo': 'Equipo', 'fecha': 'Fecha'})
# Establecemos una columna que nos indica que en los partidos de este dataframe los equipos jugaban como visitantes
df_visita['es_local'] = 0

# Unimos los partidos como local y como visitante de los equipos por separado
df_equipos = pd.concat([df_local, df_visita], ignore_index=True)


# Aseguramos que la fecha y la hora esten en el formato correcto para operar con ellas
df_equipos['Fecha'] = pd.to_datetime(df_equipos['Fecha'])
df_equipos['hora'] = pd.to_datetime(df_equipos['hora'], format='%H:%M:%S').dt.time

# Ordenamos el dataframe
df_equipos = df_equipos.sort_values(by=['Equipo', 'Fecha', 'hora']).reset_index(drop=True)

print("Columnas finales:", df_equipos.columns.tolist())
```

```python


# 1. Agrupamos y sumamos los componentes de los últimos 5 partidos (sin contar el actual)
# min_periods=1 soluciona tu problema de los NaNs iniciales
df_rolling_sums = df_equipos.groupby('Equipo')[['xg', 'tiros', 'tiros_puerta', 'pases_area', 'goles']].transform(
    lambda x: x.rolling(window=5, min_periods=1).sum().shift(1)
)
```

```python
# 1. Definimos las columnas de df_equipos que queremos promediar (Estado de forma)
cols_identificadores = ['Equipo', 'Fecha', 'hora', 'es_local', 'resultado_final']

# Seleccionamos las columnas numéricas que NO usamos en el Ratio of Sums
cols_a_promediar = [
    'xpts', 'ppda', 'corners', 'faltas', 'amarillas', 'rojas',
    'diff_xg', 'diff_tiros_puerta', 'diff_tiros', 'diff_pases_area',
    'diff_xpts', 'diff_corners', 'diff_faltas', 'diff_amarillas',
    'diff_ratio_presion', 'diff_dominio_ofensivo',
    'diff_agresividad_ofensiva', 'diff_disciplina'
]

# 2. Calculamos el Rolling Mean
df_rolling_avg = df_equipos.groupby('Equipo')[cols_a_promediar].transform(
    lambda x: x.rolling(window=5, min_periods=1).mean().shift(1)
)
df_rolling_avg.columns = [c + '_avg5' for c in df_rolling_avg.columns]




# 2. Promedio Móvil Exponencial (EWMA - Más peso a lo reciente)
# span=5 significa que mira hacia atrás unos 5 partidos de forma decreciente
df_ewma = df_equipos.groupby('Equipo')[cols_a_promediar].transform(
    lambda x: x.shift(1).ewm(span=5, adjust=False).mean()
)
df_ewma = df_ewma.add_suffix('_ewma5')


# 3. UNIÓN TOTAL
# Juntamos Identificadores + Promedios de volumen + Ratios de Suma (df_trends)
df_final_procesado = pd.concat([
    df_equipos[cols_identificadores],
    df_rolling_avg,
    df_ewma
], axis=1)

# 4. Limpieza final de NaNs (Primer partido de la liga)
df_final_procesado = df_final_procesado.fillna(df_final_procesado.median(numeric_only=True))

print("Feature Engineering Completado.")
print(f"Total columnas para el modelo: {len(df_final_procesado.columns)}")
```
